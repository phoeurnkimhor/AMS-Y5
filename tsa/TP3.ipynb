# Exercise 3.4

The data file `hours` contains monthly values of the average hours worked per week
in the U.S. manufacturing sector for July 1982 through June 1987.

1. Display and interpret the time series plot for these data.
2. Now construct a time series plot that uses separate plotting symbols for the various months. Does your interpretation change from that in part (a)?

```r
library(TSA)
data(hours)
options(repr.plot.width = 15, repr.plot.height = 10)
hours_ts <- hours

print(hours_ts)
```

```r

plot(hours_ts, ylab = "Monthly Hours", xlab = "Time", 
     main = "Average Weekly Hours in U.S. Manufacturing (1982–1987)")
```

```r
cycle(hours_ts)
```

```r
colors <- rainbow(12)
symbols <- c(16, 17, 15, 18, 3, 4, 8, 7, 9, 10, 12, 13)

plot(hours_ts, type = "n", ylab = "Monthly Hours", xlab = "Time", 
     main = "Average Weekly Hours in U.S. Manufacturing (1982–1987)",      
     cex.main = 2,   
     cex.lab = 1.2, 
     cex.axis = 2 
)
lines(hours_ts, col = "gray70", lwd = 1.5)

for (m in 1:12) {
  idx <- cycle(hours_ts) == m
  points(time(hours_ts)[idx], hours_ts[idx],
         col = colors[m],
         pch = symbols[m],
         cex = 2,
         lwd = 1.2)
}

legend("topleft", legend = month.abb, col = colors, pch = symbols, cex = 1.3)
```

# Exercise 3.5

The data file `wages` contains monthly values of the average hourly wages (in dollars)
for workers in the U.S. apparel and textile products industry for July 1981 through
June 1987.

1. Display and interpret the time series plot for these data.
2. Use least squares to fit a linear time trend to this time series. Interpret the
regression output. Save the standardized residuals from the fit for further
analysis.
3. Construct and interpret the time series plot of the standardized residuals from
part (b).
4. Use least squares to fit a quadratic time trend to the wages time series. In-
terpret the regression output. Save the standardized residuals from the fit for
further analysis.
5. Construct and interpret the time series plot of the standardized residuals from
part (d).

```r
data(wages)
# 1 
plot(wages, main='Wages', cex.main=2)
```

```r
wages_lm = lm(wages ~ time(wages))
print(summary(wages_lm))

y = rstudent(wages_lm)
sprintf("y = %f", y)
```

```r
plot(y, x=as.vector(time(wages)), type="o", xlab="Time", main="Standardized Residuals", cex.main = 2)
grid(8, 10, lty = "solid")
```

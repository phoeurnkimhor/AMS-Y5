```python
import pandas as pd
df = pd.read_csv("./elon_musk.csv", encoding="latin1")
df.head()
```

### 1. Removing URLs 
 
**Instruction**: URLs should be removed because they do not provide word-level information  
for text analysis. 

```python
df['Text'] = df['Text'].str.replace(r'http\S+|www\. \s+', '', regex=True)
df
```

### 2. Removing Punctuation  
 
**Instruction**: Remove punctuation to reduce complexity, as it typically does not hold vital  
information. 

```python
df['Text'] = df['Text'].str.translate(str.maketrans('', '', string.punctuation))
df
```

### 3. Removing Stop Words 
 
**Instruction**: Remove stopwords (common words like articles and prepositions) because they  
add little to no value to the meaning of the sentence. 

```python
common_stopwords = pd.read_csv("./common_stopwords.txt")
stop_words = list(common_stopwords['STOP_WORDS'])
pattern = r'\b(' + '|'.join(stop_words) + r')\b'
df['Text'] = df['Text'].str.replace(pattern, '', regex=True)
df
```

```python
pattern
```

### 4. Lowercasing 
 
**Instruction**: Lowercase your text to improve general accuracy, as many language models  
treat capital letters differently. It is generally recommended for an English corpus.

```python
df['Text'] = df['Text'].str.lower()
df
```

### 5. Tokenization 
 
**Instruction**: This is the foundational step in IR-the process of breaking your text (corpus)  
into individual units called tokens (by Sentence, by Paragraph, or by word). 

```python
from nltk.tokenize import sent_tokenize, word_tokenize

df['word_tokens'] = df['Text'].apply(word_tokenize)
df['sentence_tokens'] = df['Text'].apply(sent_tokenize)
```

### 6. Stemming (Reduction to Root Word) 
 
**Instruction**: Use stemming to reduce inflected words to their word stem. 
Method: It is a heuristic process that cuts off the ends of words. 

```python
from nltk.stem.snowball import SnowballStemmer
stemmer = SnowballStemmer("english")

df['word_stemming'] = df['Text'].apply(lambda x: [stemmer.stem(word) for word in word_tokenize(x)])
df
```

### 7. Lemmatization (Reduction to Dictionary Form) 
 
**Instruction**: Use lemmatization to reduce inflected words to their base or dictionary form,  
the lemma. 
Method: It uses vocabulary and morphological analysis. 

```python
from nltk.stem import WordNetLemmatizer
# nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()

df['lemmatization'] = df["Text"].apply(lambda x: [lemmatizer.lemmatize(word) for word in word_tokenize(x)])
df
```
